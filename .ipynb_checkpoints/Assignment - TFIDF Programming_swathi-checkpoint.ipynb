```python
#2. Import Libraries
```

```python
import os
import nltk
import re
import pandas as pd

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

nltk.download('punkt')
nltk.download('stopwords')
```

```python
#3. Load the Documents
```

```python
path = "Data"

documents = {}

for file in os.listdir(path):
    with open(os.path.join(path, file), "r", encoding="utf8") as f:
        documents[file] = f.read().lower()

print(documents.keys())
```

```python
#Step 1 — Tokenization
```

```python
#4. Tokenization = splitting text into words.
```

```python
tokens = {}

for doc, text in documents.items():
    words = word_tokenize(text)
    tokens[doc] = words
```

```python
#counting tokens and types
```

```python
for doc, words in tokens.items():
    print(doc)
    print("Tokens:", len(words))
    print("Unique Tokens:", len(set(words)))
    print()
```

```python
#corpus totals
all_tokens = [tok for doc_tokens in tokens.values() for tok in doc_tokens]

print("Total tokens in corpus:", len(all_tokens))
print("Total unique tokens/types in corpus:", len(set(all_tokens)))
```

```python
#Step 2 - Stopword Removal
```

```python
#5. Stopwords are common words like the, is, and, in.
```

```python
stop_words = set(stopwords.words('english'))

clean_tokens = {}

for doc, words in tokens.items():
    filtered = [w for w in words if w.lower() not in stop_words and w.isalpha()]
    clean_tokens[doc] = filtered
```

```python
#counting again
```

```python
for doc, words in clean_tokens.items():
    print(doc)
    print("Tokens after stopwords:", len(words))
    print("Unique words:", len(set(words)))
    print()
```

```python
#Step 3 — Stemming
#Convert words to their root form.
```

```python
stemmer = PorterStemmer()

stemmed_docs = {}

for doc, words in clean_tokens.items():
    stemmed = [stemmer.stem(w) for w in words]
    stemmed_docs[doc] = stemmed
```

```python
vocab = set()

for words in stemmed_docs.values():
    vocab.update(words)

print("Total Vocabulary Size:", len(vocab))
```

```python
#terms left in each document after stemming
for doc, words in stemmed_docs.items():
    print(doc)
    print("Terms after stemming (unique):", len(set(words)))
    print()
```

```python
#Step 4 - TF-IDF Calculation

#6. Now convert documents into numerical vectors.
```

```python
texts = [" ".join(words) for words in stemmed_docs.values()]

vectorizer = TfidfVectorizer(
    tokenizer=str.split,
    preprocessor=None,
    token_pattern=None,
    lowercase=False
)

tfidf_matrix = vectorizer.fit_transform(texts)

print(tfidf_matrix.shape)
```

```python
#Step 5 - Cosine Similarity

#7.Now compute similarity between documents.
```

```python
similarity = cosine_similarity(tfidf_matrix)
```

```python
#8. convert to Table
```

```python
doc_names = list(stemmed_docs.keys())

similarity_df = pd.DataFrame(similarity,
                             index=doc_names,
                             columns=doc_names)

similarity_df
```
