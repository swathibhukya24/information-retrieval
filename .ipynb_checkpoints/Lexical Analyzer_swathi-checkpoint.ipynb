```python
import re

def lexical_analyzer(input_string):
    token_specification = [
        ('INTEGER',       r'\b\d+\b'),
        ('ASSIGN_OP',     r'='),
        ('INEQUALITY',    r'!=|==|<>|>'),
        ('PLUS',          r'\+'),
        ('MINUS',         r'-'),
        ('MULTIPLY',      r'\*'),
        ('DIVIDE',        r'/'),
        ('LPAREN',        r'\('),
        ('RPAREN',        r'\)'),
        ('SEMICOLON',     r';'),
        ('IDENTIFIER',    r'\b[a-zA-Z_][a-zA-Z0-9_]*\b'),
        ('SKIP',          r'[ \t]+'),
        ('MISMATCH',      r'.'),
    ]

    tok_regex = '|'.join('(?P<%s>%s)' % pair for pair in token_specification)
    tokens = []

    for mo in re.finditer(tok_regex, input_string):
        kind = mo.lastgroup
        value = mo.group()
        if kind == 'SKIP':
            continue
        elif kind == 'MISMATCH':
            print(f"Unexpected character: {value}")
        else:
            tokens.append((value, kind))

    return tokens
```

```python
input_code = "index = 2 * (count + 17);"
tokens = lexical_analyzer(input_code)

print("Lexeme\t\tToken")
print("-----------------------------")
for lexeme, token in tokens:
    print(f"{lexeme}\t\t{token}")
```

```python
input_code = "muSum = 2 * count + 17;"
tokens = lexical_analyzer(input_code)

print("Lexeme\t\tToken")
print("-----------------------------")
for lexeme, token in tokens:
    print(f"{lexeme}\t\t{token}")
```
